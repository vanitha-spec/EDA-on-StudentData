```python
import pandas as pd
```

```python
def clean_whitespace(df):
    """
    Remove unnecessary spaces from text columns.
    """

    result = df.copy()

    result["Name"] = result["Name"].str.strip()
    result["Gender"] = result["Gender"].str.strip()
    result["City"] = result["City"].str.strip()

    return result
```

```python
def standardize_categories():
    result = df.copy()
    result["City"]=result["City"].replace(
        {
            "hyderabad": "Hyderabad",
            "HYDERABAD": "Hyderabad",

            "chennai": "Chennai",
            "CHENNAI": "Chennai",

            "bengaluru": "Bengaluru",
            "BENGALURU": "Bengaluru",

            "mumbai": "Mumbai",
            "MUMBAI": "Mumbai",
        }
    )
    result["Gender"] = result["Gender"].replace({
        "M": "Male",
        "m": "Male",
        "MALE": "Male",

        "F": "Female",
        "f": "Female",
        "FEMALE": "Female",
    })
    return result
```

```python
def convert_numeric_columns(df):
    result = df.copy()
    result["Age"] = pd.to_numeric(result["Age"],
    errors="coerce"
    )
    result["Marks"] = pd.to_numeric(result["Marks"],
    errors="coerce"
    )
    return result
```

```python
def remove_duplicates(df):
    result = df.copy()
    result = result.drop_duplicates()
    return result
```

```python
def handle_missing_values(df):
    result = df.copy()
    result= result.dropna(subset=["Name"])

    result.loc[
        (result["Age"]>1 | result["Age"]>100),"Age"
    ] = pd.NA

    result.loc[
        (result["Age"]<1 | result["Age"]>100), "Age"
    ] = pd.NA
    result["Marks"]=result["Marks"].fillna(
        result["Marks"].mean()
    )
    return result
```
